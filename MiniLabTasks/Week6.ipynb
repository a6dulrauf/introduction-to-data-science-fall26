```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(100)
```

```python
n_people = 100000
is_sick = np.random.choice([0, 1], size=n_people, p=[0.99, 0.01])
```

```python
pos_test = np.zeros(n_people)

for i in range(n_people):
  if is_sick[i] == 1:
    pos_test[i] = np.random.choice([1, 0], p=[0.90, 0.10])
  else:
    pos_test[i] = np.random.choice([1, 0], p=[0.10, 0.90])
```

```python
df = pd.DataFrame({'is_sick': is_sick, 'pos_test': pos_test})
```

```python
positives = df[df['pos_test'] == 1]
share_sick = (positives['is_sick'] == 1).mean()
print(f'Share of sick people among those who test positive: {share_sick:.4f}')
```
